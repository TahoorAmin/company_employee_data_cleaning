# Company Employee Data Cleaning 

# Objective:

The objective of this assignment is to clean and prepare a messy company employee dataset using Python and Pandas. The dataset will be inspected to identify missing values, inconsistent entries, incorrect data types, and duplicate records. Appropriate data cleaning techniques will then be applied to improve the quality and consistency of the dataset. Finally, the cleaned dataset will be verified, compared with the original dataset, and exported as a CSV file.

```python
import pandas as pd
df=pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")
```

```python
df_cleaned=df.copy()
```

Before cleaning the dataset, we inspect its structure, dimensions, columns, data types, and statistical summary to understand the dataset and identify potential data quality issues.

# DATA CLEANING

# 1. Handling Incorrect Data Types

The data types of the columns are checked to identify any inappropriate data types. The Joining Date column is stored as a string/object type and needs to be converted to datetime format for proper date handling and analysis.

```python
df_cleaned["Joining_Date"]=pd.to_datetime(df_cleaned["Joining_Date"])
```

```python
df_cleaned["Joining_Date"].dtype
```

# 2. Identifying and Handling Data Quality Issues: Missing Values and Inconsistent Entries

The dataset is examined column by column to identify missing values and inconsistent entries. Based on the nature of each column, appropriate cleaning techniques are applied to resolve the identified issues while preserving as much useful data as possible.

```python
df.isna().sum()
```

```python
df["Department"].unique()
```

```python
#Handling the inconsistent entries

df_cleaned["Department"]=df_cleaned["Department"].replace({
    "ENGINEERING":"Engineering" ,
    "engineering":"Engineering"
})
```

```python
df_cleaned["Department"].unique()
```

```python
#Filling the missing values with mode

df_cleaned["Department"]=df_cleaned["Department"].fillna(df_cleaned["Department"].mode()[0])
```

```python
df["Age"].mean().round(0)

```

```python
#Filling the missing values with mean

df_cleaned["Age"]=df_cleaned["Age"].fillna(df_cleaned["Age"].mean()).round(0)
```

```python
df["Gender"].unique()
```

```python
#Handling the inconsistent entries

df_cleaned["Gender"]=df_cleaned["Gender"].replace({
    "MALE":"Male" ,
    "female":"Female"
})
```

```python
df_cleaned["Gender"].unique()
```

```python
#Filling the missing values with mode

df_cleaned["Gender"]=df_cleaned["Gender"].fillna(df_cleaned["Gender"].mode()[0])
```

```python
df["Annual_Salary"].mean()
```

```python
#Filling the missing values with mean

df_cleaned["Annual_Salary"]=df_cleaned["Annual_Salary"].fillna(df_cleaned["Annual_Salary"].mean())
```

```python
df["Experience_Years"].mean().round(1)
```

```python
#Filling the missing values with mean

df_cleaned["Experience_Years"]=df_cleaned["Experience_Years"].fillna(df_cleaned["Experience_Years"].mean().round(1))
```

```python
df["City"].unique()
```

```python
#Handling the inconsistent entries

df_cleaned["City"]=df_cleaned["City"].replace({
    "delhi":"Delhi" ,
    "DELHI":"Delhi",
    "Delhi ":"Delhi",

})
```

```python
df_cleaned["City"].unique()
```

```python
#Filling the missing values with mode

df_cleaned["City"]=df_cleaned["City"].fillna(df_cleaned["City"].mode()[0])
```

```python
df["Performance_Score"].unique()
```

```python
#Filling the missing values with mode

df_cleaned["Performance_Score"]=df_cleaned["Performance_Score"].fillna(df_cleaned["Performance_Score"].mode()[0])
```

```python
df["Work_Mode"].unique()
```

```python
#Handling the inconsistent entries

df_cleaned["Work_Mode"]=df_cleaned["Work_Mode"].replace({
    "REMOTE":"Remote" ,
    "remote":"Remote"
})
```

```python
df_cleaned["Work_Mode"].unique()
```

```python
#Filling the missing values with mode

df_cleaned["Work_Mode"]=df_cleaned["Work_Mode"].fillna(df_cleaned["Work_Mode"].mode()[0])
```

```python
df["Job_Title"].unique()
```

# Verification After Cleaning

The dataset is rechecked after handling the identified missing values and inconsistent entries. The results show that there are no remaining missing values or unresolved inconsistencies in the cleaned dataset.

```python
df_cleaned.isna().sum()
```

# 3. Handling Duplicate Records

The cleaned dataset is checked for duplicate records. Any duplicate rows identified are removed to ensure that each employee record appears only once.

```python
df.duplicated().sum()
```

```python
df_cleaned=df_cleaned.drop_duplicates()
```

```python
df_cleaned.duplicated().sum()
```

# Before and After Cleaning Comparison

The original and cleaned datasets are compared to evaluate the changes made during the data cleaning process. The comparison focuses on the number of rows, missing values, duplicate records, and other relevant data quality improvements.

```python
comparison = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Total Missing Values",
        "Duplicate Records"
    ],
    "Before Cleaning": [
        len(df),
        len(df.columns),
        df.isna().sum().sum(),
        df.duplicated().sum()
    ],
    "After Cleaning": [
        len(df_cleaned),
        len(df_cleaned.columns),
        df_cleaned.isna().sum().sum(),
        df_cleaned.duplicated().sum()
    ]
})

comparison
```

- Identified and handled missing values using appropriate imputation methods such as mean and mode.
- Standardized inconsistent entries in categorical columns.
- Identified and removed duplicate records from the dataset.
- Verified the cleaned dataset to ensure that no missing values or duplicate records remained.
- Compared the dataset before and after cleaning to evaluate the improvements.
